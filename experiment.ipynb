```python
from sklearn.datasets import make_classification
from sklearn.linear_model import SGDClassifier
from sklearn.model_selection import train_test_split

N_SAMPLES = 10_000
baseline_data = make_classification(n_samples=N_SAMPLES, n_features=4, n_classes=2, return_X_y=True, random_state=42)
X_fit, X_cal, y_fit, y_cal = train_test_split(
    baseline_data[0], baseline_data[1], test_size=0.5, random_state=42
)
LogisticRegression_model = SGDClassifier(max_iter=200)

LogisticRegression_model.fit(X_fit, y_fit)
score = LogisticRegression_model.score(X_cal, y_cal)

threshold = 0.4 * (score + LogisticRegression_model.score(X_fit, y_fit))
threshold, score
```
