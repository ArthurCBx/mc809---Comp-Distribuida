```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df_train = pd.read_csv('results/ff1.0_alpha100.0//train.csv', names=['server_round', 'cid', 'acc', 'loss'])
df_test  = pd.read_csv('results/ff1.0_alpha100.0//evaluate.csv', names=['server_round', 'cid', 'acc', 'loss'])

fig, ax = plt.subplots(1, 2, figsize=(15, 5))
fig.suptitle("Resultados do Treinamento e Teste para o modelo FedAvg com alpha=100.0 e frac=1.0")

sns.lineplot(data=df_train, x='server_round', y='loss', ax=ax[0], color='b', label='Loss Treino')
sns.lineplot(data=df_train, x='server_round', y='acc', ax=ax[1], color='b', label='Acc Treino')
sns.lineplot(data=df_test, x='server_round', y='loss', ax=ax[0], color='r', label='Loss Teste')
sns.lineplot(data=df_test, x='server_round', y='acc', ax=ax[1], color='r', label='Acc Teste')

ax[0].set_title('Loss')
ax[1].set_title('Accuracy')

ax[0].legend()
ax[1].legend()

ax[0].set_xticks(range(0, 21, 5))
ax[1].set_xticks(range(0, 21, 5))


ax[0].grid(True, linestyle=':')
ax[1].grid(True, linestyle=':')
```

```python
df_train_server_f1_alpha100  = pd.read_csv('results/ff1.0_alpha100.0/server.csv', names=['server_round', 'acc'])
df_train_server_f1_alpha1    = pd.read_csv('results/ff1.0_alpha1.0/server.csv', names=['server_round', 'acc'])
df_train_server_f1_alpha01   = pd.read_csv('results/ff1.0_alpha0.1/server.csv', names=['server_round', 'acc'])
df_train_server_f05_alpha100 = pd.read_csv('results/ff0.5_alpha100.0/server.csv', names=['server_round', 'acc'])
df_train_server_f05_alpha1   = pd.read_csv('results/ff0.5_alpha1.0/server.csv', names=['server_round', 'acc'])
df_train_server_f05_alpha01  = pd.read_csv('results/ff0.5_alpha0.1/server.csv', names=['server_round', 'acc'])
df_train_server_f05_iid      = pd.read_csv('results/ff0.5_iid/server.csv', names=['server_round', 'acc'])

plt.figure(figsize=(10, 5))
sns.lineplot(data=df_train_server_f1_alpha100, x='server_round', y='acc', color='b', label='f1_alpha100 Treino')
sns.lineplot(data=df_train_server_f1_alpha1, x='server_round', y='acc', color='g', label='f1_alpha1 Treino')
sns.lineplot(data=df_train_server_f1_alpha01, x='server_round', y='acc', color='r', label='f1_alpha01 Treino')
sns.lineplot(data=df_train_server_f05_alpha100, x='server_round', y='acc', color='c', label='f05_alpha100 Treino')
sns.lineplot(data=df_train_server_f05_alpha1, x='server_round', y='acc', color='m', label='f05_alpha1 Treino')
sns.lineplot(data=df_train_server_f05_alpha01, x='server_round', y='acc', color='y', label='f05_alpha01 Treino')
sns.lineplot(data=df_train_server_f05_iid, x='server_round', y='acc', color='k', label='f05_iid Treino')
plt.title("Comparação de Acurácia do Servidor para diferentes valores de alpha e frac")
plt.xlabel("Rodada do Servidor")
plt.ylabel("Acurácia")
plt.xticks(range(0, 21, 5))
plt.legend()
plt.grid(True)
plt.show()
```
